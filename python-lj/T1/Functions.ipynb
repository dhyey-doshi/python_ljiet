```python
#ADDITION (without arg without return)
def Add():
    a=10
    b=20
    c=a+b
    print(c)
Add()
```

```python
#SUBSTRACTION (without arg with return)
def Sub():
    a=10
    b=20
    return b-a
c=Sub()
print(c)
```

```python
#MULTIPLICATION (with arg without return)
def Mul(a,b):
    print(a*b)
Mul(5,2)
```

```python
#DIVISION (with arg with return)
def Div(a,b):
    return a/b
c=Div(10,2)
print(c)
```

```python
#MAST QOSTEN
net_amt=int(input("pls enter total amount"))
def break_amt(net):
    fifties=net//50
    temp1=net%50
    twenties=temp1//20
    temp2=temp1%20
    tens=temp2//10
    temp3=temp2%10
    fives=temp3//5
    temp4=temp3%5
    twos=temp4//2
    temp5=temp4%2
    ones=temp5//1
    if fifties!=0:
        print(f"{fifties} coins of 50")
    if twenties!=0:
        print(f"{twenties} coins of 20")
    if tens!=0:
        print(f"{tens} coins of 10")
    if fives!=0:
        print(f"{fives} coins of 5")
    if twos!=0:
        print(f"{twos} coins of 2")
    if ones!=0:
        print(f"{ones} coins of 1")
    #print(f"{fifties} coins of 50, {twenties} coins of 20, {tens} coins of 10, {fives} coins of 5, {twos} coins of 2, {ones} coins of 1")
break_amt(net_amt)
```
